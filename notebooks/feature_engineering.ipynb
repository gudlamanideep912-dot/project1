```python
# ============================================================
# CELL 1 - PROJECT SETUP
# ============================================================

from pathlib import Path
import sys
import pandas as pd

# Project root
PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("=" * 70)
print("PHISHGUARD AI - FEATURE ENGINEERING")
print("=" * 70)

print("\nProject root:")
print(PROJECT_ROOT)

print("\nProject root exists:")
print(PROJECT_ROOT.exists())
```

```python
# ============================================================
# CELL 2 - IMPORT URL FEATURE EXTRACTOR
# ============================================================

from src.features.url_features import (
    extract_url_features,
    extract_features_from_dataframe
)

print(
    "URL feature extractor imported successfully. ✅"
)
```

```python
# ============================================================
# CELL 3 - LOAD CLEANED DATASET
# ============================================================

CLEANED_DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "cleaned_phishing_dataset.csv"
)

print("Loading:")
print(CLEANED_DATA_PATH)

if not CLEANED_DATA_PATH.exists():
    raise FileNotFoundError(
        f"Cleaned dataset not found:\n"
        f"{CLEANED_DATA_PATH}"
    )

cleaned_dataset = pd.read_csv(
    CLEANED_DATA_PATH
)

print("\nDataset loaded successfully. ✅")

print("\nShape:")
print(cleaned_dataset.shape)

print("\nColumns:")
print(
    cleaned_dataset.columns.tolist()
)
```

```python
# ============================================================
# CELL 4 - CHECK REQUIRED COLUMNS
# ============================================================

print("=" * 70)
print("CHECKING DATASET COLUMNS")
print("=" * 70)

print("\nAvailable columns:")

for i, column in enumerate(
    cleaned_dataset.columns,
    start=1
):
    print(f"{i:2}. {column}")

# ------------------------------------------------------------
# Find URL column
# ------------------------------------------------------------

possible_url_columns = [
    "URL",
    "url",
    "Url",
    "url_string",
    "URL_STRING"
]

URL_COLUMN = None

for column in possible_url_columns:
    if column in cleaned_dataset.columns:
        URL_COLUMN = column
        break

if URL_COLUMN is None:
    raise KeyError(
        "Could not find the URL column."
    )

# ------------------------------------------------------------
# Find label column
# ------------------------------------------------------------

possible_label_columns = [
    "label",
    "Label",
    "LABEL",
    "status",
    "Status",
    "target",
    "Target"
]

TARGET_COLUMN = None

for column in possible_label_columns:
    if column in cleaned_dataset.columns:
        TARGET_COLUMN = column
        break

if TARGET_COLUMN is None:
    raise KeyError(
        "Could not find the target/label column."
    )

print("\nURL column:")
print(URL_COLUMN)

print("\nTarget column:")
print(TARGET_COLUMN)

print("\nRequired columns found successfully. ✅")
```

```python
# ============================================================
# CELL 5 - TEST URL FEATURE EXTRACTION
# ============================================================

TEST_URL = "https://www.google.com"

test_features = extract_url_features(
    TEST_URL
)

print("=" * 70)
print("URL FEATURE TEST")
print("=" * 70)

print("\nURL:")
print(TEST_URL)

print("\nNumber of features:")
print(len(test_features))

print("\nFeature names:")

for i, feature_name in enumerate(
    test_features.keys(),
    start=1
):
    print(
        f"{i:2}. {feature_name}"
    )

print("\nFeature values:")

for feature_name, feature_value in test_features.items():
    print(
        f"{feature_name:40} : {feature_value}"
    )
```

```python
# ============================================================
# CELL 6 - EXTRACT FEATURES FROM DATASET
# ============================================================

print("=" * 70)
print("EXTRACTING URL FEATURES")
print("=" * 70)

feature_rows = extract_features_from_dataframe(
    cleaned_dataset,
    url_column=URL_COLUMN
)

url_features_df = pd.DataFrame(
    feature_rows
)

print("\nFeature extraction completed. ✅")

print("\nShape:")
print(url_features_df.shape)

print("\nNumber of features:")
print(len(url_features_df.columns))

print("\nFirst 5 rows:")

display(
    url_features_df.head()
)
```

```python
# ============================================================
# CELL 7 - VERIFY FEATURES
# ============================================================

feature_columns = list(
    url_features_df.columns
)

print("=" * 70)
print("FEATURE VERIFICATION")
print("=" * 70)

print("\nNumber of features:")
print(len(feature_columns))

print("\nFeature columns:")

for i, column in enumerate(
    feature_columns,
    start=1
):
    print(f"{i:2}. {column}")

if len(feature_columns) != 25:
    raise RuntimeError(
        f"Expected 25 features, "
        f"but found {len(feature_columns)}."
    )

print(
    "\n25 URL features confirmed. ✅"
)
```

```python
# ============================================================
# CELL 8 - ADD TARGET LABEL
# ============================================================

url_features_df[TARGET_COLUMN] = (
    cleaned_dataset[TARGET_COLUMN].values
)

print("=" * 70)
print("TARGET LABEL ADDED")
print("=" * 70)

print("\nFinal dataset shape:")
print(url_features_df.shape)

print("\nFeature count:")
print(
    len(url_features_df.columns) - 1
)

print("\nTarget column:")
print(TARGET_COLUMN)

print("\nTarget distribution:")
print(
    url_features_df[TARGET_COLUMN]
    .value_counts()
)

print(
    "\nTarget label added successfully. ✅"
)
```

```python
# ============================================================
# CELL 9 - CHECK MISSING VALUES
# ============================================================

print("=" * 70)
print("MISSING VALUE CHECK")
print("=" * 70)

missing_values = (
    url_features_df.isnull()
    .sum()
)

total_missing = (
    missing_values.sum()
)

print("\nTotal missing values:")
print(total_missing)

if total_missing == 0:

    print(
        "\nNo missing values found. ✅"
    )

else:

    print(
        "\nMissing values found: ⚠️"
    )

    print(
        missing_values[
            missing_values > 0
        ]
    )
```

```python
# ============================================================
# CELL 10 - SAVE FEATURE DATASET
# ============================================================

PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FEATURE_DATA_PATH = (
    PROCESSED_DIR
    / "phishing_features.csv"
)

url_features_df.to_csv(
    FEATURE_DATA_PATH,
    index=False
)

print("=" * 70)
print("FEATURE DATASET SAVED")
print("=" * 70)

print("\nPath:")
print(FEATURE_DATA_PATH)

print("\nFile exists:")
print(FEATURE_DATA_PATH.exists())

if FEATURE_DATA_PATH.exists():

    print("\nFile size:")
    print(
        FEATURE_DATA_PATH.stat().st_size,
        "bytes"
    )

    print(
        "\nFeature dataset saved successfully. ✅"
    )
```

```python
# ============================================================
# CELL 11 - VERIFY SAVED DATASET
# ============================================================

verification_df = pd.read_csv(
    FEATURE_DATA_PATH
)

print("=" * 70)
print("SAVED DATASET VERIFICATION")
print("=" * 70)

print("\nShape:")
print(verification_df.shape)

print("\nColumns:")
print(
    verification_df.columns.tolist()
)

print("\nFeature count:")
print(
    len(verification_df.columns) - 1
)

if len(verification_df.columns) - 1 != 25:
    raise RuntimeError(
        "Saved dataset does not contain "
        "25 feature columns."
    )

print(
    "\nSaved dataset verified successfully. ✅"
)
```

```python
# ============================================================
# CELL 12 - SAVE FEATURE COLUMN ORDER
# ============================================================

import json

FEATURE_COLUMNS_PATH = (
    PROJECT_ROOT
    / "models"
    / "feature_columns.json"
)

FEATURE_COLUMNS_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

with open(
    FEATURE_COLUMNS_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        feature_columns,
        file,
        indent=4
    )

print("=" * 70)
print("FEATURE COLUMN ORDER SAVED")
print("=" * 70)

print("\nPath:")
print(FEATURE_COLUMNS_PATH)

print("\nNumber of features:")
print(len(feature_columns))

print(
    "\nFeature column order saved successfully. ✅"
)
```

```python
# ============================================================
# CELL 13 - PREPARE TRAIN / TEST DATA
# ============================================================

from sklearn.model_selection import train_test_split

X = url_features_df[feature_columns].copy()
y = url_features_df[TARGET_COLUMN].copy()

print("=" * 70)
print("TRAIN / TEST PREPARATION")
print("=" * 70)

print("\nX shape:")
print(X.shape)

print("\ny shape:")
print(y.shape)

print("\nNumber of features:")
print(X.shape[1])

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining data:")
print(X_train.shape)

print("\nTesting data:")
print(X_test.shape)

print("\nTrain/test split completed successfully. ✅")
```

```python
# ============================================================
# CELL 14 - TRAIN RANDOM FOREST
# ============================================================

from sklearn.ensemble import RandomForestClassifier

rf_model_new = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

print("=" * 70)
print("TRAINING RANDOM FOREST")
print("=" * 70)

rf_model_new.fit(
    X_train,
    y_train
)

print("\nRandom Forest training completed. ✅")

print("\nNumber of trees:")
print(rf_model_new.n_estimators)

print("\nNumber of features:")
print(rf_model_new.n_features_in_)
```

```python
# ============================================================
# CELL 15 - RANDOM FOREST EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

y_pred = rf_model_new.predict(
    X_test
)

y_probability = rf_model_new.predict_proba(
    X_test
)[:, 1]

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    y_probability
)

print("=" * 70)
print("RANDOM FOREST EVALUATION")
print("=" * 70)

print(f"\nAccuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1-Score  : {f1 * 100:.2f}%")
print(f"ROC-AUC   : {roc_auc * 100:.2f}%")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_pred
    )
)
```

```python
# ============================================================
# CELL 16 - SAVE NEW RANDOM FOREST
# ============================================================

import joblib

NEW_MODEL_PATH = (
    PROJECT_ROOT
    / "models"
    / "random_forest_25_features.joblib"
)

joblib.dump(
    rf_model_new,
    NEW_MODEL_PATH
)

print("=" * 70)
print("NEW RANDOM FOREST SAVED")
print("=" * 70)

print("\nPath:")
print(NEW_MODEL_PATH)

print("\nExists:")
print(NEW_MODEL_PATH.exists())

if NEW_MODEL_PATH.exists():
    print("\nSize:")
    print(
        f"{NEW_MODEL_PATH.stat().st_size / 1024:.2f} KB"
    )

    print(
        "\nModel saved successfully. ✅"
    )
```
