```python
# Cell 1: Import required libraries

from pathlib import Path
import pandas as pd
import numpy as np

print("Libraries imported successfully. ✅")
```

```python
# Cell 2: Define project paths

PROJECT_ROOT = Path.cwd().parent

RAW_DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "PhiUSIIL_Phishing_URL_Dataset.csv"
)

PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)

PROCESSED_DATA_PATH = (
    PROCESSED_DIR
    / "cleaned_dataset.csv"
)

print("Project root:")
print(PROJECT_ROOT)

print("\nRaw dataset:")
print(RAW_DATA_PATH)

print("\nProcessed dataset:")
print(PROCESSED_DATA_PATH)
```

```python
# Cell 3: Verify input dataset

if not RAW_DATA_PATH.exists():

    raise FileNotFoundError(
        f"Dataset not found:\n{RAW_DATA_PATH}\n\n"
        "Make sure the CSV is inside data/raw/"
    )

print("PhiUSIIL dataset found. ✅")
```

```python
# Cell 4: Load the raw dataset

df = pd.read_csv(
    RAW_DATA_PATH,
    low_memory=False
)

print("Dataset loaded successfully. ✅")
print(f"Rows    : {len(df):,}")
print(f"Columns : {len(df.columns):,}")
```

```python
# Cell 5: Create a working copy

clean_df = df.copy()

print("Working copy created. ✅")
```

```python
# Cell 6: Validate required columns

REQUIRED_COLUMNS = {
    "URL",
    "label"
}

missing_columns = (
    REQUIRED_COLUMNS
    - set(clean_df.columns)
)

if missing_columns:

    raise ValueError(
        "Required columns are missing: "
        f"{sorted(missing_columns)}"
    )

print("Required columns found. ✅")
print("URL")
print("label")
```

```python
# Cell 7: Record original dataset size

original_rows = len(clean_df)
original_columns = len(clean_df.columns)

print(f"Original rows    : {original_rows:,}")
print(f"Original columns : {original_columns:,}")
```

```python
# Cell 8: Analyze missing values

missing_values = (
    clean_df
    .isnull()
    .sum()
    .sort_values(
        ascending=False
    )
)

missing_values = missing_values[
    missing_values > 0
]

if missing_values.empty:

    print("No missing values found. ✅")

else:

    print("Missing values:")
    print(missing_values)
```

```python
# Cell 9: Remove rows missing critical fields

before = len(clean_df)

clean_df = clean_df.dropna(
    subset=["URL", "label"]
).copy()

removed = before - len(clean_df)

print(f"Rows removed: {removed:,}")
print(f"Rows remaining: {len(clean_df):,}")
```

```python
# Cell 10: Clean URL values

clean_df["URL"] = (
    clean_df["URL"]
    .astype(str)
    .str.strip()
)

print("URL whitespace cleaned. ✅")
```

```python
# Cell 11: Remove empty URL records

before = len(clean_df)

clean_df = clean_df[
    clean_df["URL"].str.len() > 0
].copy()

removed = before - len(clean_df)

print(f"Empty URLs removed: {removed:,}")
print(f"Rows remaining: {len(clean_df):,}")
```

```python
# Cell 12: Convert labels to numeric

clean_df["label"] = pd.to_numeric(
    clean_df["label"],
    errors="coerce"
)

print(
    clean_df["label"]
    .value_counts(dropna=False)
)
```

```python
# Cell 13: Remove invalid labels

before = len(clean_df)

clean_df = clean_df[
    clean_df["label"].isin([0, 1])
].copy()

removed = before - len(clean_df)

print(f"Invalid label rows removed: {removed:,}")
```

```python
# Cell 14: Set final label data type

clean_df["label"] = (
    clean_df["label"]
    .astype("int8")
)

print(
    "Label data type:",
    clean_df["label"].dtype
)
```

```python
# Cell 15: Remove duplicate records

before = len(clean_df)

clean_df = (
    clean_df
    .drop_duplicates()
    .reset_index(drop=True)
)

removed = before - len(clean_df)

print(f"Duplicate rows removed: {removed:,}")
print(f"Rows remaining: {len(clean_df):,}")
```

```python
# Cell 16: Check duplicate URLs

duplicate_urls = (
    clean_df["URL"]
    .duplicated()
    .sum()
)

print(
    f"Duplicate URL entries: "
    f"{duplicate_urls:,}"
)
```

```python
# Cell 17: Final label distribution

label_counts = (
    clean_df["label"]
    .value_counts()
    .sort_index()
)

print("Final label distribution")
print("=" * 40)

for label, count in label_counts.items():

    name = (
        "Phishing"
        if label == 0
        else "Legitimate"
    )

    print(
        f"{label} ({name}): "
        f"{count:,}"
    )
```

```python
# Cell 18: Final missing-value check

remaining_missing = (
    clean_df
    .isnull()
    .sum()
)

remaining_missing = remaining_missing[
    remaining_missing > 0
]

if remaining_missing.empty:

    print(
        "No missing values remain. ✅"
    )

else:

    print(
        "Remaining missing values:"
    )

    print(remaining_missing)
```

```python
# Cell 19: Check final data types

print(
    clean_df.dtypes.to_string()
)
```

```python
# Cell 20: Final dataset validation

assert "URL" in clean_df.columns
assert "label" in clean_df.columns

assert clean_df["URL"].notna().all()
assert clean_df["URL"].str.len().gt(0).all()

assert clean_df["label"].isin([0, 1]).all()

assert not clean_df.empty

print("All validation checks passed. ✅")
```

```python
# Cell 21: Cleaning summary

final_rows = len(clean_df)

rows_removed = (
    original_rows - final_rows
)

print("=" * 70)
print("PHISHGUARD AI - DATA CLEANING SUMMARY")
print("=" * 70)

print(
    f"Original rows : {original_rows:,}"
)

print(
    f"Final rows    : {final_rows:,}"
)

print(
    f"Rows removed  : {rows_removed:,}"
)

print(
    f"Columns       : {len(clean_df.columns):,}"
)

print(
    f"Missing values: "
    f"{clean_df.isnull().sum().sum():,}"
)

print(
    f"Duplicates    : "
    f"{clean_df.duplicated().sum():,}"
)

print("\nCleaning completed successfully. ✅")
```

```python
# Cell 22: Create processed-data directory

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print(
    "Processed directory ready. ✅"
)
```

```python
# Cell 23: Save cleaned dataset

clean_df.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print(
    "Cleaned dataset saved successfully. ✅"
)

print(
    f"\nSaved to:\n{PROCESSED_DATA_PATH}"
)
```

```python
# Cell 24: Verify processed dataset

if PROCESSED_DATA_PATH.exists():

    file_size_mb = (
        PROCESSED_DATA_PATH.stat().st_size
        / (1024 * 1024)
    )

    print("Processed dataset verified. ✅")
    print(
        f"File size: {file_size_mb:.2f} MB"
    )

else:

    raise FileNotFoundError(
        "Processed dataset was not created."
    )
```
