```python
# ============================================================
# CELL 16 - EXTRACT FEATURES FROM COMPLETE DATASET
# ============================================================

feature_rows = extract_features_from_dataframe(
    cleaned_dataset,
    url_column="URL"
)

url_features_df = pd.DataFrame(
    feature_rows
)

print("=" * 70)
print("URL FEATURE DATASET CREATED")
print("=" * 70)

print("\nShape:")
print(url_features_df.shape)

print("\nNumber of features:")
print(len(url_features_df.columns))

print("\nFeature columns:")
for number, column in enumerate(
    url_features_df.columns,
    start=1
):
    print(f"{number:2}. {column}")

print("\nFirst 5 rows:")
display(url_features_df.head())
```

```python
# ============================================================
# CELL 17 - ADD LABEL
# ============================================================

url_features_df["label"] = (
    cleaned_dataset["label"].values
)

print("=" * 70)
print("LABEL ADDED")
print("=" * 70)

print("\nFinal shape:")
print(url_features_df.shape)

print("\nColumns:")
print(list(url_features_df.columns))

print("\nLabel distribution:")
print(
    url_features_df["label"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# CELL 18 - SAVE FEATURE DATASET
# ============================================================

PROCESSED_DIR = (
    PROJECT_ROOT
    / "data"
    / "processed"
)

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FEATURES_PATH = (
    PROCESSED_DIR
    / "phishing_features.csv"
)

url_features_df.to_csv(
    FEATURES_PATH,
    index=False
)

print("=" * 70)
print("FEATURE DATASET SAVED SUCCESSFULLY ✅")
print("=" * 70)

print("\nPath:")
print(FEATURES_PATH)

print("\nFile exists:")
print(FEATURES_PATH.exists())

print("\nFile size:")
print(
    FEATURES_PATH.stat().st_size,
    "bytes"
)
```
