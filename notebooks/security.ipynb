```python
# ============================================================
# CELL 1 - PHISHGUARD SECURITY MODULE
# ============================================================

from pathlib import Path
import sys
import re

PROJECT_ROOT = Path.cwd().resolve()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("=" * 70)
print("PHISHGUARD AI - SECURITY MODULE")
print("=" * 70)

print("\nProject root:")
print(PROJECT_ROOT)

print("\nSecurity module initialized successfully. ✅")
```

```python
# ============================================================
# CELL 2 - URL SAFETY VALIDATION
# ============================================================

from urllib.parse import urlparse


def validate_url(url):
    """
    Validate a URL before sending it to the ML model.
    """

    if not isinstance(url, str):
        return False, "URL must be a string."

    url = url.strip()

    if not url:
        return False, "URL cannot be empty."

    if len(url) > 2048:
        return False, "URL is too long."

    try:
        parsed = urlparse(url)
    except Exception:
        return False, "Invalid URL format."

    if parsed.scheme not in ["http", "https"]:
        return False, "Only HTTP and HTTPS URLs are supported."

    if not parsed.netloc:
        return False, "URL does not contain a valid hostname."

    return True, "URL is structurally valid."


print("URL validation function created successfully. ✅")
```

```python
# ============================================================
# CELL 3 - TEST URL VALIDATION
# ============================================================

test_urls = [
    "https://www.google.com",
    "https://example.com/login",
    "",
    "not-a-url",
    "ftp://example.com"
]

print("=" * 70)
print("URL VALIDATION TEST")
print("=" * 70)

for url in test_urls:

    valid, message = validate_url(url)

    print("\nURL:", url)
    print("Valid:", valid)
    print("Message:", message)
```

```python
# ============================================================
# CELL 4 - SECURITY CHECKS
# ============================================================

def security_check(url):
    """
    Perform basic security checks before prediction.
    """

    valid, message = validate_url(url)

    if not valid:
        return {
            "safe_to_analyze": False,
            "message": message
        }

    parsed = urlparse(url)

    hostname = parsed.hostname or ""

    # Basic hostname sanity checks
    if ".." in hostname:
        return {
            "safe_to_analyze": False,
            "message": "Hostname contains consecutive dots."
        }

    if hostname.startswith(".") or hostname.endswith("."):
        return {
            "safe_to_analyze": False,
            "message": "Hostname format is invalid."
        }

    return {
        "safe_to_analyze": True,
        "message": "URL passed security checks."
    }


print(
    "Security check function created successfully. ✅"
)
```

```python
# ============================================================
# CELL 5 - SECURITY CHECK TEST
# ============================================================

TEST_URL = "https://www.google.com"

security_result = security_check(
    TEST_URL
)

print("=" * 70)
print("SECURITY CHECK")
print("=" * 70)

print("\nURL:")
print(TEST_URL)

print("\nSafe to analyze:")
print(
    security_result["safe_to_analyze"]
)

print("\nMessage:")
print(
    security_result["message"]
)

print(
    "\nSecurity check completed successfully. ✅"
)
```

```python
# ============================================================
# CELL 6 - SECURE ANALYSIS WRAPPER
# ============================================================

def secure_analyze_url(url, analysis_function):
    """
    Validate and security-check a URL before
    passing it to the PhishGuard analysis function.
    """

    security_result = security_check(url)

    if not security_result["safe_to_analyze"]:

        return {
            "success": False,
            "url": url,
            "message": security_result["message"]
        }

    try:

        result = analysis_function(url)

        return {
            "success": True,
            "url": url,
            "message": "Analysis completed successfully.",
            "analysis": result
        }

    except Exception as error:

        return {
            "success": False,
            "url": url,
            "message": (
                "Analysis failed safely. "
                f"Reason: {error}"
            )
        }


print(
    "Secure analysis wrapper created successfully. ✅"
)
```

```python
# ============================================================
# CELL 7 - FINAL SECURITY TEST
# ============================================================

TEST_URL = "https://www.google.com"

print("=" * 70)
print("PHISHGUARD AI - SECURE ANALYSIS TEST")
print("=" * 70)

print("\nURL:")
print(TEST_URL)

print(
    "\nSecurity validation:"
)

security_result = security_check(
    TEST_URL
)

print(
    security_result
)

print(
    "\nSecurity notebook completed successfully. ✅"
)
```
